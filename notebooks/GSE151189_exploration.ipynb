# GSE151189 — Dataset Exploration

## Dataset Overview

- **Organism:** *Plasmodium falciparum*
- **Total samples:** 156
- **Platform:** GPL18893
- **Study:** Global RNA expression profiling of K13-edited mutant and isogenic wild-type parasites
- **PubMed IDs:** 33483501, 38632343
- **Technology:** Microarray

## Parasite Lines

Five genetically edited parasite lines were investigated:

- Dd2 wild type
- Dd2 R539T
- Dd2 C580Y
- Cam3.II reverted wild type
- Cam3.II R539T

## Experimental Design

The study investigated:

1. Basal gene expression during the 48-hour intra-erythrocytic developmental cycle.
2. Transcriptional responses to dihydroartemisinin (DHA).
3. Differences between K13-mutant and corresponding wild-type parasites.

DHA treatment conditions:

- DHA concentration: 700 nM
- Vehicle control: 0.1% DMSO
- Three independent experiments for each strain and condition, except Cam3.II DHA treatment, which was performed once.

## Developmental-Stage Information

- Dd2 lines: 0, 3, 6, 16, 24, 32 and 48 hours.
- Cam3.II lines: 0, 8, 16, 24, 32, 40 and 48 hours.

The time-course design provides information about transcriptional changes across the IDC.

## Limitations

- The dataset contains controlled laboratory parasite lines rather than independent clinical isolates.
- Cam3.II DHA-treated groups have only one experiment.
- Dd2 and Cam3.II have different sampling schedules.
- Microarray expression data require appropriate preprocessing before comparative analysis.

## Screening Decision

**Retain — strong mechanistic comparison candidate.**

GSE151189 is highly relevant because it compares K13-mutant and isogenic wild-type parasites across developmental stages and under DHA exposure. It can help investigate resistance-associated transcriptional changes and developmental-stage effects.

However, it should not be treated as independent clinical validation.
