```python
!pip -q install GEOparse

import GEOparse
import pandas as pd

gse61536 = GEOparse.get_GEO(
    "GSE61536",
    destdir="data/raw",
    how="full"
)

print("GSE ID:", gse61536.name)
print("Number of samples:", len(gse61536.gsms))
print("Platforms:", list(gse61536.gpls.keys()))
```

```python
sample_records = []

for gsm_name, gsm in gse61536.gsms.items():
    record = {
        "GSM": gsm_name,
        "Title": gsm.metadata.get("title", [""])[0],
        "Source": gsm.metadata.get("source_name_ch1", [""])[0],
        "Characteristics": " | ".join(
            gsm.metadata.get("characteristics_ch1", [])
        ),
        "Description": " | ".join(
            gsm.metadata.get("description", [])
        )
    }
    sample_records.append(record)

gse61536_summary = pd.DataFrame(sample_records)

pd.set_option("display.max_colwidth", 150)
display(gse61536_summary)
```

```python
for gsm_name, gsm in gse61536.gsms.items():
    title = gsm.metadata.get("title", [""])[0]
    characteristics = gsm.metadata.get("characteristics_ch1", [])

    print(f"{gsm_name}: {title}")
    print("  ", " | ".join(characteristics))
```

```python
for key in ["title", "summary", "overall_design", "type", "pubmed_id"]:
    print(f"\n{key.upper()}:")
    print("\n".join(gse61536.metadata.get(key, ["Not available"])))
```

# GSE61536 — Dataset Exploration

## Dataset Overview

- **Organism:** *Plasmodium falciparum*
- **Strain:** K1
- **Total samples:** 29
- **Platform:** GPL13818
- **Study:** Plasmodium parasites mount an arrest response against dihydroartemisinin
- **PubMed ID:** 26490244
- **Experiment:** DHA treatment versus vehicle control

## Experimental Design

| Timepoint | Vehicle control | DHA-treated |
|---|---:|---:|
| 1 hour | 5 | 5 |
| 2 hours | 5 | 5 |
| 3 hours | 4 | 5 |
| **Total** | **14** | **15** |

## Main Findings Reported by the Study

- DHA-treated trophozoites showed a more ring-like transcriptional response.
- Cytosolic ribosomal protein genes were notably downregulated.
- DHA-treated rings showed a more muted response.
- Schizonts were reported to be unresponsive to DHA.
- The study suggests DHA can induce developmental arrest.

## Limitations

- Only one *P. falciparum* strain (K1) is represented in the inspected samples.
- No direct resistant-versus-sensitive isolate comparison.
- The experiment investigates drug response rather than resistance-associated signatures.
- The samples are experimental treatment replicates, not independent clinical isolates.

## Screening Decision

**Retain as a supplementary dataset.**

GSE61536 may help interpret DHA-induced transcriptional changes and developmental arrest. However, it is not suitable as a primary independent validation dataset for comparing artemisinin-resistance signatures.
