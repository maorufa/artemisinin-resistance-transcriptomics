```python
!pip -q install GEOparse

import GEOparse
import pandas as pd

gse121505 = GEOparse.get_GEO(
    "GSE121505",
    destdir="data/raw",
    how="full"
)

print("GEO ID:", gse121505.name)
print("Number of samples:", len(gse121505.gsms))
print("Platforms:", list(gse121505.gpls.keys()))
```

```python
sample_records = []

for gsm_name, gsm in gse121505.gsms.items():
    record = {
        "GSM": gsm_name,
        "Title": gsm.metadata.get("title", [""])[0],
        "Source": gsm.metadata.get("source_name_ch1", [""])[0],
        "Characteristics": " | ".join(
            gsm.metadata.get("characteristics_ch1", [])
        ),
        "Description": " | ".join(
            gsm.metadata.get("description", [])
        )
    }
    sample_records.append(record)

gse121505_summary = pd.DataFrame(sample_records)

pd.set_option("display.max_colwidth", 150)
display(gse121505_summary)
```

# GSE121505 — Dataset Exploration

## 1. Dataset Overview

- **GEO accession:** GSE121505
- **Organism:** *Plasmodium falciparum*
- **Total samples:** 12
- **Platform:** GPL15130
- **Parasite strain:** 3D7
- **Transgenic line:** HP1
- **Study type:** Expression profiling by array

## 2. Sample Characteristics

The dataset contains 12 samples representing gametocyte development from Day 1 to Day 12.

| Characteristic | Observation |
|---|---|
| Parasite strain | 3D7 |
| Transgenic line | HP1 |
| Developmental period | Day 1–Day 12 |
| Number of samples | 12 |
| Artemisinin treatment groups | Not identified |
| Resistance groups | Not identified |

## 3. Relevance to the Research

GSE121505 provides a gametocyte developmental expression reference. It may help distinguish gametocyte-associated transcriptional patterns from expression signatures observed in asexual blood-stage parasites.

However, this dataset does not directly investigate artemisinin resistance or DHA response.

## 4. Limitations

- Only 12 samples are available.
- The dataset represents gametocyte development rather than the asexual intra-erythrocytic developmental cycle.
- No artemisinin-resistant and sensitive comparison groups are identified.
- Biological replication is not established from the inspected metadata.
- Its platform differs from GPL18893 used by several other screened datasets.

## 5. Screening Decision

**Retain as a supplementary reference only.**

GSE121505 will not be included in the primary resistance-signature comparison. It may provide additional biological context when interpreting gametocyte-associated expression.
