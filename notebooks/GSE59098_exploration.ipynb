```python
# Install required packages
!pip -q install GEOparse pandas numpy

import GEOparse
import pandas as pd
import numpy as np

# Download GSE59098
gse = GEOparse.get_GEO(
    "GSE59098",
    destdir="data/raw",
    how="brief"
)

print("DATASET: GSE59098")
print("=" * 45)

print("Total GEO samples:", len(gse.gsms))
print("Platform IDs:", list(gse.gpls.keys()))

# Inspect the first five samples
sample_records = []

for gsm_id, gsm in list(gse.gsms.items())[:5]:

    sample_records.append({
        "GSM_ID": gsm_id,
        "Title": gsm.metadata.get("title", [""])[0],
        "Characteristics": gsm.metadata.get(
            "characteristics_ch1", []
        )
    })

sample_preview = pd.DataFrame(sample_records)

display(sample_preview)
```

```python
import re
from collections import Counter

sample_records = []

for gsm_id, gsm in gse.gsms.items():

    title = gsm.metadata.get("title", [""])[0]

    characteristics = gsm.metadata.get(
        "characteristics_ch1", []
    )

    # Convert characteristics into a readable dictionary
    char_dict = {}

    for item in characteristics:
        if ":" in item:
            key, value = item.split(":", 1)
            char_dict[key.strip().lower()] = value.strip()

    # Extract isolate ID and timepoint from the title
    parts = title.split("-")

    isolate_id = "-".join(parts[:2]) if len(parts) >= 3 else None

    timepoint = re.search(r"(\d+)h", title)

    sample_records.append({
        "GSM_ID": gsm_id,
        "Title": title,
        "Isolate_ID": isolate_id,
        "Timepoint": timepoint.group(1) + "h" if timepoint else None,
        "Culture": char_dict.get("culture"),
        "Stage": char_dict.get("stage")
    })

gse59098_summary = pd.DataFrame(sample_records)

print("Total GEO samples:", len(gse59098_summary))
print("Unique isolates:", gse59098_summary["Isolate_ID"].nunique())

print("\nSamples per timepoint:")
display(
    gse59098_summary["Timepoint"]
    .value_counts(dropna=False)
    .sort_index()
    .to_frame("Number of samples")
)

print("\nSamples per isolate (first 15):")
display(
    gse59098_summary["Isolate_ID"]
    .value_counts()
    .head(15)
    .to_frame("Number of samples")
)

print("\nSample metadata preview:")
display(gse59098_summary.head(10))
```

```python
# Count measurements at each timepoint for every isolate

isolate_timepoint_table = pd.crosstab(
    gse59098_summary["Isolate_ID"],
    gse59098_summary["Timepoint"]
)

print("Number of isolates:", len(isolate_timepoint_table))

print("\nMeasurements per isolate:")
display(isolate_timepoint_table)

print("\nCompleteness:")
print(
    "Isolates with all 6 timepoints:",
    (isolate_timepoint_table == 1).all(axis=1).sum()
)

print(
    "Isolates with missing or repeated timepoints:",
    (~(isolate_timepoint_table == 1).all(axis=1)).sum()
)
```

```python
# Inspect all metadata fields for resistance-related information

all_characteristics = []

for gsm_id, gsm in gse.gsms.items():

    for item in gsm.metadata.get("characteristics_ch1", []):
        all_characteristics.append(item)

unique_characteristics = sorted(set(all_characteristics))

for item in unique_characteristics:
    if any(word in item.lower() for word in [
        "clearance", "resistance", "artemisinin",
        "half-life", "ic50", "susceptibility",
        "drug", "phenotype"
    ]):
        print(item)
```

```python
# Check clearance half-life consistency across repeated samples

clearance_records = []

for gsm_id, gsm in gse.gsms.items():

    title = gsm.metadata.get("title", [""])[0]
    isolate_id = "-".join(title.split("-")[:2])

    characteristics = gsm.metadata.get(
        "characteristics_ch1", []
    )

    for item in characteristics:

        if "clearance halflife" in item.lower():

            value = item.split(":", 1)[-1].strip()

            clearance_records.append({
                "Isolate_ID": isolate_id,
                "GSM_ID": gsm_id,
                "Clearance_Half_Life": value
            })

clearance_df = pd.DataFrame(clearance_records)

print("Samples with clearance information:",
      len(clearance_df))

print("Unique isolates with clearance information:",
      clearance_df["Isolate_ID"].nunique())

print("\nNumber of clearance values per isolate:")
display(
    clearance_df.groupby("Isolate_ID")[
        "Clearance_Half_Life"
    ].nunique().value_counts()
)

print("\nClearance phenotype by isolate:")
display(
    clearance_df.groupby("Isolate_ID")[
        "Clearance_Half_Life"
    ].first().to_frame()
)
```

```python
# Check whether expression data are available in GEO samples

sample_table_info = []

for gsm_id, gsm in gse.gsms.items():

    table = gsm.table

    sample_table_info.append({
        "GSM_ID": gsm_id,
        "Rows": table.shape[0],
        "Columns": table.shape[1],
        "Has_expression_data": not table.empty
    })

expression_availability = pd.DataFrame(sample_table_info)

print("Samples with expression data:",
      expression_availability["Has_expression_data"].sum())

print("Total samples:",
      len(expression_availability))

print("\nExpression table dimensions:")
display(
    expression_availability[
        ["Rows", "Columns"]
    ].describe()
)

print("\nFirst five samples:")
display(expression_availability.head())
```

```python
# Check expression tables after full download

print("Total samples:", len(gse_full.gsms))

loaded_samples = 0

for gsm_id, gsm in gse_full.gsms.items():
    if not gsm.table.empty:
        loaded_samples += 1

print("Samples with loaded expression tables:", loaded_samples)

# Inspect one sample
first_gsm_id = next(iter(gse_full.gsms))
first_gsm = gse_full.gsms[first_gsm_id]

print("\nExample GSM:", first_gsm_id)
print("Table shape:", first_gsm.table.shape)

display(first_gsm.table.head())
```

### GSE59098 — Dataset Screening

**Decision: Retain as a candidate**

- Organism: *Plasmodium falciparum*
- Experimental design: Ex vivo time-course study
- Total GEO samples: 110
- Independent isolates: 19
- Sampling timepoints: 0, 8, 16, 24, 32 and 40 hours
- Complete time courses: 15/19 isolates
- Resistance phenotype: Parasite-clearance half-life following artemisinin treatment
- Clearance half-life range: 3.437–7.815 hours
- Expression data: Available for all 110 samples
- Expression table dimensions: 11,005 probes × 2 columns per sample (`ID_REF`, `VALUE`)
- Platform: GPL18893
- Potential research role: Developmental-stage-aware analysis of resistance-associated transcriptional signatures.

**Limitations:** Only 19 independent isolates; four have incomplete or repeated timepoints. Sampling time does not directly establish developmental stage. GSE59098 is part of the same broader study as GSE59097 and should not be considered an independent validation dataset.

**Screening conclusion:** Retain for further analysis.
